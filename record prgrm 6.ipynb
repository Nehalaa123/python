```python
n=int(input("Enter a number:"))
print("Factors are:")
for i in range(1,n+1):
    if n%i==0:
        print(i)
```

```python
words=input("Enter words").split()
max_len=0
for word in words:
    if len(word)>max_len:
        max_len=len(word)
print("Length of longest word:",max_len)        
        
```
