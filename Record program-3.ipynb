```python
names=["Anu","Arun","Nayana","Aswathi"]
count=0
for name in names:
    count=count+name.lower().count('a')
print("Occurence of a:",count)
```

```python
color_list1=["Yellow","Blue","purble","Black"]
color_list2=["Blue","Black","Red"]
for color in color_list1:
    if color not in color_list2:
        print(color)   
```

```python
d={4:"Mango",2:"Pappaya",1:"Apple",3:"Orange"}
print("Ascending order:")
for key in sorted(d):
    print(key,":",d[key])
print("Descending order:")
for key in sorted(d,reverse=True):
    print(key,":",d[key])
                
```
